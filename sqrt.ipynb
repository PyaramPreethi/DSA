```python
def sqrt(x):

    left = 1
    right = x
    ans = 0
    if x < 2:
        return x
    while left <= right:
        mid = (left+right)//2
        if mid * mid == x:
            return mid
        elif mid * mid < x:
            ans = mid
            left = mid+1
        else:
            right = mid - 1

    return ans
x =78
print("num : ",sqrt(x))
```
