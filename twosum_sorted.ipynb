```python
def twosum_sorted(nums,target):
    
    left = 0
    right = len(nums)-1
    while  left < right:
        sums = nums[left]+nums[right]
        if sums == target:
            return [nums[left],nums[right]]
    
        elif sums < target:
            left+=1
        else:
            right-=1

    return[]
nums = [1,4,5,7,8,10,19]
target = 15    
print(twosum_sorted(nums,target))
```
