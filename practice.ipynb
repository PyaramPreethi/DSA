```python
def move_zeroes(nums):
    inser_pos = 0
    for i in range(len(nums)):
        if nums[i]!=0:
            nums[inser_pos] = nums[i]
            inser_pos+=1

    for i in range(inser_pos,len(nums)):
        nums[i]=0
    return nums








nums = [1,2,0,3,4,0,4,0,9,8,0,8]
print("length",move_zeroes(nums))
```

```python
def two_sum(nums,target):
    nums.sort()
    left = 0
    right = len(nums)-1
    while left < right:
        sum = nums[left]+nums[right]
        if sum == target:
            return [nums[left],nums[right]]

        elif sum < target:
            left +=1

        elif sum > target:
            right -=1
    return []

target = 15
nums = [56,8,10,7,8,15]
print("twosums",two_sum(nums,target))
```
