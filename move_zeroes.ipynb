```python
def move_zeroes(nums):
    insert_pos = 0
    for i in range(len(nums)):
        if nums[i]!=0:
            nums[insert_pos] = nums[i]
            insert_pos+=1
    for i in range(insert_pos,len(nums)):
        nums[i] = 0
    return nums
nums = [1,6,8,0,8,9,0,6,4,0,9]
print("Sorted",move_zeroes(nums))
```
